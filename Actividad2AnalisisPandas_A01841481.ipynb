4. Imprime nuevamente las cantidad de valores únicos por columna.
- ¿Por qué son 144 horas diferentes?
- La cantidad de combinaciones únicas de fecha y hora no coincide con el número total de mediciones. Aunque no haya valores NaN, esto no garantiza que no se hayan omitido registros. Identifica las fechas en las que faltan mediciones.

Aunque el dataframe no presenta valores NaN, esto no garantiza que estén registradas todas las observaciones esperadas. Una medición puede estar ausente porque no existe el registro completo correspondiente a una determinada combinación de fecha y hora. Por esta razón, se agrupó la información por fecha y se contabilizaron los valores únicos de hora, considerando que un día completo debe contener 144 mediciones. Mediante este procedimiento se identificaron las fechas que presentan menos de las 144 mediciones esperadas y, por tanto, aquellas en las que existen mediciones faltantes.

Una vez concluida esta verificación, se conservaron únicamente los registros correspondientes al año 2020, de acuerdo con el período de análisis establecido para el conjunto de datos, y se eliminó la columna medicion, manteniendo las columnas fecha y hora para los análisis posteriores.

8. Utiliza el dataframe de indicadores diarios para crear un nuevo dataframe `indicadores_mensuales`. Para ello:
- Crea una nueva columna `mes` a partir de la fecha que se encuentra en el índice.
- Haciendo uso de `groupby()`, calcula los promedios mensuales y conserva únicamente las columnas `humedad_relativa`, `temperatura_celsius` y `velocidad_viento` en el nuevo dataframe.

```python
# Convertir el índice a formato datetime
indicadores_diarios.index = pd.to_datetime(indicadores_diarios.index)

# Crear la columna mes a partir de la fecha del índice
indicadores_diarios['mes'] = indicadores_diarios.index.month

# Calcular los promedios mensuales de las columnas solicitadas
indicadores_mensuales = indicadores_diarios.groupby('mes')[
    ['humedad_relativa', 'temperatura_celsius', 'velocidad_viento']
].mean()

# Mostrar el nuevo dataframe
display(indicadores_mensuales)
```

9. Construye el mismo dataframe mensual, pero esta vez utilizando la función `pivot_table()`.
- Añade la columna `indice_calor` usando la fórmula de Thom (*Temperature - Humidity Index*, TDI):
> `temperatura_celsius + 0.33 x humedad_relativa - 0.7 x velocidad_viento - 4`

```python
# Construir el dataframe mensual utilizando pivot_table()
indicadores_mensuales = indicadores_diarios.pivot_table(
    index='mes',
    values=['humedad_relativa', 'temperatura_celsius', 'velocidad_viento'],
    aggfunc='mean'
)

# Añadir la columna indice_calor
indicadores_mensuales['indice_calor'] = (
    indicadores_mensuales['temperatura_celsius']
    + 0.33 * indicadores_mensuales['humedad_relativa']
    - 0.7 * indicadores_mensuales['velocidad_viento']
    - 4
)

display(indicadores_mensuales)
```

10. Recuerda que de las cuatro funciones estudiadas para manipular la estructura del dataframe:
- `melt/pivot` hacen las transformaciones de manera controlada, es decir puedes
definir por medio de sus parámetros qué variables quedarán como índices, columnas y valores.
- `stack/unstack` la conversión se aplica siempre sobre los niveles inferiores de index/columns.
- Convierte el dataframe del ejercicio anterior a formato largo usando: `melt()` y `stack()` y comenta las diferencias.

```python
# Construir el dataframe mensual utilizando pivot_table()
indicadores_mensuales = indicadores_diarios.pivot_table(
    index='mes',
    values=['humedad_relativa', 'temperatura_celsius', 'velocidad_viento'],
    aggfunc='mean'
)

# Añadir la columna indice_calor
indicadores_mensuales['indice_calor'] = (
    indicadores_mensuales['temperatura_celsius']
    + 0.33 * indicadores_mensuales['humedad_relativa']
    - 0.7 * indicadores_mensuales['velocidad_viento']
    - 4
)

display(indicadores_mensuales)
```

Ambos métodos permiten convertir el dataframe a formato largo, pero lo hacen de manera diferente. Con melt() la transformación es más controlada, ya que permite especificar qué variable se conserva como identificador y cómo se llamarán las columnas que contienen los indicadores y sus valores. En este caso, mes se mantiene como identificador y los demás datos se organizan en indicador y valor.

En cambio, stack() realiza la transformación directamente sobre el nivel inferior de las columnas y genera una estructura con índices jerárquicos. Por tanto, melt() ofrece mayor control sobre la organización y los nombres del resultado, mientras que stack() realiza la conversión basándose en la estructura de índices y columnas del dataframe.
